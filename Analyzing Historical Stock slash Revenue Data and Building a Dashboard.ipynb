```python
import yfinance as yf

tesla = yf.Ticker("GME")
data = tesla.history().reset_index()

data.head()
```

```python
import matplotlib.pyplot as plt

def make_graph(x, y, title):
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='Close Price', color='blue')
    plt.xlabel('Date')
    plt.ylabel('Price (USD)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

# Use the function
make_graph(
    x=data['Date'],
    y=data['Close'],
    title='GameStop Stock Price Over Time'
)
```
